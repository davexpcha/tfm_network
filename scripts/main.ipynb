# Setup

```python
import numpy as np
from dataclasses import dataclass, field
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression

SOURCE_TRAINING_SET = "data/UNSW_NB15_training-set(in).csv"
SOURCE_TEST_SET = "data/UNSW_NB15_testing-set(in).csv"
SOURCE_SEL = [x for x in range(45) if x not in {0, 43}] # ignoring Y-category

@dataclass
class OneHotEncoding:
    m: dict[str, float] = field(default_factory=dict)

    def encode(self, s) -> float:
        if s not in self.m:
            self.m[s] = float(2 ** len(self.m))
        return self.m[s]

PROTO_CONV = OneHotEncoding()
SERVICE_CONV = OneHotEncoding()
STATE_CONV = OneHotEncoding()

def get_data(path):
    return np.loadtxt(
        path,
        delimiter=",",
        skiprows=1,
        usecols=SOURCE_SEL,
        converters={
            2: PROTO_CONV.encode,
            3: SERVICE_CONV.encode,
            4: STATE_CONV.encode,
        },
    )
```

## Generate regression model

```python
TRAINING_DATA = get_data(SOURCE_TRAINING_SET)
X_train_raw = TRAINING_DATA[:, :42]
Y_train = TRAINING_DATA[:, 42:43]

X_train = StandardScaler().fit_transform(X_train_raw)

reg = LinearRegression().fit(X_train, Y_train)

reg
```

## Test regression model generated

```python
TEST_DATA = get_data(SOURCE_TEST_SET)

X_test_raw = TEST_DATA[:, :42]
Y_test = TEST_DATA[:, 42:43]

X_test = StandardScaler().fit_transform(X_test_raw)

result = reg.score(X_test, Y_test)
result
```
